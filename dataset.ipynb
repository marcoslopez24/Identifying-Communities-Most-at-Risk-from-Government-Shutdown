# Importing Datasets

```python
import pandas as pd
import requests
```

## FARA - Food Access Research Atlas
We are interested in the <code>LILATracts_1And10</code> and <code>HUNVFlag</code> variables.

```python
FARA = pd.read_csv('./2019 Food Access Research Atlas Data/Food Access Research Atlas.csv')
```

```python
FARA.head()
```

```python
FARA_FD = FARA[['CensusTract', 'State', 'County', 'Urban', 'LILATracts_1And10', 'HUNVFlag']].copy()
FARA_FD.head()
```

## SNAP - SNAP County Level and Participation Data
We are interested in the <code>Calc: SNAP Total PA and Non-PA People</code> column.

```python
SNAP = pd.read_excel('./snap-zip-fns388a-7/Jan 2019.xls', header=3)
```

```python
SNAP.columns
```

```python
SNAP = SNAP.drop(columns=['Unnamed: 2', 'Unnamed: 4', 'Unnamed: 6', 'Unnamed: 8', 'Unnamed: 10', 'Unnamed: 12'])
SNAP.head()
```

```python
SNAP_PR = SNAP[['Substate/Region', 'Calc: SNAP Total PA and Non-PA People']].copy()
SNAP_PR.head()
```

## ACS - American Community Survey
We are interested in these data tables: <code>S1901_C01_012E</code>, <code>S1701_C03_001E</code>, <code>S0101_C01_001E</code>.
- <code>S1901_C01_012E</code>: Median household income (inflation-adjusted, 2010 dollars)
- <code>S1701_C03_001E</code>: Percent of people below poverty level
- <code>S0101_C01_001E</code>: Total population

```python
with open('ACS_key.txt', 'r') as file:
    key = file.read()
key = key.rstrip('\n')
```

```python
api_url = f'https://api.census.gov/data/2019/acs/acs5/subject?get=NAME,S1901_C01_012E,S1701_C03_001E,S0101_C01_001E&for=county:*&key={key}'
response = requests.get(api_url)
```

```python
ACS = pd.DataFrame(response.json())
ACS.columns = ACS.iloc[0]
ACS = ACS.drop(0).reset_index(drop=True)
ACS.head()
```

## Joining For Final Dataset

First we aggregate the tract level food atlas data to county level and extract fips

```python
FARA_FD['fips'] = FARA_FD['CensusTract'].astype(str).str[:5]

FARA_FD.head()
```

```python
def mode(x):
    m = x.mode()
    return m.iloc[0] if not m.empty else None

FARA_FD = FARA_FD.groupby(['State', 'County']).agg({
    'Urban': 'mean',
    'LILATracts_1And10': 'mean',
    'HUNVFlag': 'mean',
    'fips': mode
}).reset_index()

FARA_FD
```

Now we extract fips for <code>SNAP</code>

```python
SNAP_PR['fips'] = SNAP_PR['Substate/Region'].str.extract(r'(^\d{5})')
SNAP_PR = SNAP_PR.rename(columns={'Calc: SNAP Total PA and Non-PA People': 'SNAP_Participants'})
SNAP_PR = SNAP_PR[SNAP_PR['fips'].notna()]

SNAP_PR
```

Now extract fips for <code>ACS</code>

```python
ACS['fips'] = ACS['state'] + ACS['county']
ACS.head()
```

Now we merge

```python
# --- Food Access (FARA): keep fips + food access + optional labels ---
FARA_small = FARA_FD[['fips', 'State', 'County', 'Urban', 'LILATracts_1And10', 'HUNVFlag']]

# --- SNAP: keep fips + total participants ---
SNAP_small = SNAP_PR[['fips', 'SNAP_Participants']]

# --- ACS: keep fips + income + poverty + population ---
ACS_small = ACS[['fips', 'S1901_C01_012E', 'S1701_C03_001E', 'S0101_C01_001E']].rename(
    columns={
        'S1901_C01_012E': 'MedianIncome',
        'S1701_C03_001E': 'PovertyPercent',
        'S0101_C01_001E': 'TotalPopulation'
    }
)
```

```python
FARA_small.head()
```

```python
SNAP_small.head()
```

```python
ACS_small.head()
```

```python
acs_snap = ACS_small.merge(SNAP_small, on='fips', how='inner')
final_df = acs_snap.merge(FARA_small, on='fips', how='inner')
final_df = final_df[[
    'fips',
    'State',
    'County',
    'Urban',
    'LILATracts_1And10',
    'HUNVFlag',
    'SNAP_Participants',
    'MedianIncome',
    'PovertyPercent',
    'TotalPopulation'
]]

final_df.head()
```

```python
final_df.to_csv('final_food_access_analysis_dataset.csv', index=False)
```
